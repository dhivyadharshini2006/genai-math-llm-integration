```python
from dotenv import load_dotenv,find_dotenv
```

```python
openai.api_key = os.environ['OPENAI_API_KEY']
```

```python
import json

def check_hotel(city):
    """Check hotel availability."""
    hotels = {
        "Chennai": "15 rooms available",
        "Mumbai": "8 rooms available",
        "Hyderabad": "12 rooms available"
    }

    return json.dumps({
        "city": city,
        "availability": hotels.get(city, "No hotels found")
    })
```

```python
functions = [
    {
        "name": "check_hotel",
        "description": "Check hotel room availability in a given city",
        "parameters": {
            "type": "object",
            "properties": {
                "city": {
                    "type": "string",
                    "description": "The name of the city, e.g. Chennai, Mumbai, Hyderabad"
                }
            },
            "required": ["city"]
        }
    }
]
```

```python
messages = [
    {
        "role": "user",
        "content": "What are the hotels in Chennai ?"
    }
]
```

```python
response = openai.ChatCompletion.create(
    model="gpt-3.5-turbo",
    messages=messages,
    functions=functions
)
```

```python
print(response)
```

```python
response_message = response["choices"][0]["message"]
```

```python
check_hotel(args["city"])
```
